## 1. 二维森林：自由走动 vs 一条铁轨（长文 §2）

把边长 L 的正方形切成单位格子。自由搜索 ≈ L×L 个格子；一条穿过它的轨 ≈ L 个格子。


```python
print(f"{'L':>6} {'平面格子 L²':>16} {'轨道格子 L':>14} {'平面/轨道':>12}")
print("-" * 52)
for L in (2, 5, 10, 50, 100):
    plane, rail = L * L, L
    print(f"{L:>6} {plane:>16,} {rail:>14,} {plane/rail:>12.0f}×")

print("\n结论：边长变 10 倍，平面选项 ×100，轨道只 ×10。LoRA 就是主动走轨道。")
```

## 2. 维数一高，全空间先淹死你（长文 §2 / §5）

每边 10 个采样点。d 维超立方体的格点数 = 10^d。人眼只到 3 维；一层 4096×4096 的权重是一千六百万维。


```python
per_axis = 10
print(f"每轴 {per_axis} 个采样点：\n")
print(f"{'维数 d':>8} {'格点数 10^d':>24}  图像")
print("-" * 56)
labels = {
    1: "铁轨",
    2: "森林地面",
    3: "立方体",
    8: "最小网四点的自由微调（下一站）",
    20: "4×5 一层全参更新（环节 04）",
}
for d in (1, 2, 3, 8, 20):
    n = per_axis ** d
    print(f"{d:>8} {n:>24,}  {labels[d]}")

print("\n20 维、每轴 10 点已经是 10^20 个格子。")
print("全参微调不是「多调几个数」，是搜索体积爆炸。")
```

## 3. 铺 r 条轨：旋钮从 mn 降到 r(m+n)（长文 §3）

一层 m×n。全参 mn 个旋钮；LoRA 秩 r 字面 r(m+n) 个（环节 04 再减 r² 的真自由度）。


```python
def count(m, n, r):
    full = m * n
    lora = r * (m + n)
    return full, lora, full / lora

cases = [
    ("视频 4×5，秩 1", 4, 5, 1),
    ("视频 4×5，秩 2", 4, 5, 2),
    ("最小网 4×2，秩 1", 4, 2, 1),
    ("4096×4096，秩 8", 4096, 4096, 8),
    ("4096×4096，秩 16", 4096, 4096, 16),
    ("4096×4096，秩 64", 4096, 4096, 64),
]
print(f"{'场景':<22} {'全参 mn':>14} {'LoRA r(m+n)':>14} {'省了':>10}")
print("-" * 64)
for name, m, n, r in cases:
    full, lora, ratio = count(m, n, r)
    print(f"{name:<22} {full:>14,} {lora:>14,} {ratio:>9.1f}×")

print("\n4×5 秩 2 已经 18 vs 20——轨快铺满森林，低秩优势消失。")
print("所以 r 必须 ≪ min(m,n)。4096 维上 r=16 仍省 128 倍（单投影）。")
```
