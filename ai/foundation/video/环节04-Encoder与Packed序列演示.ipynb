```python
# 极小序列：2 个视频格子 + 3 个文本 + 2 个音频
tags = [0, 0, 1, 1, 1, 2, 2]  # video=0 text=1 audio=2
# 视频/音频可能有不同 timestep 索引；这里视频用 t_id=0，音频用 t_id=1
t_of_row = [0, 0, 0, 0, 0, 1, 1]


def combined_index(t_id, tag):
    return t_id * 3 + max(tag, 0)


print("row  tag  t_id  combined  含义")
names = {0: "vid", 1: "txt", 2: "aud"}
for i, (tag, tid) in enumerate(zip(tags, t_of_row)):
    c = combined_index(tid, tag)
    print(f"{i:3d}  {names[tag]:3}  {tid:4}  {c:8}  第{tid}套时间步的{names[tag]}调制")

print("\n投影维：视频96→5376，音频32→5376，文本5120→5376")
print("自己 cat 却对不上 t_of_row，就会把音频的 shift 套到画面上。")
```
