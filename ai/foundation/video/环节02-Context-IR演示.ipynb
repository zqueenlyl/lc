> 把含糊创意拆成关系三元组，体会 IR 在压缩什么。零依赖。


```python
raw = "参考视频1的希区柯克镜头，让图2的人唱歌，歌声参考音频3。"

# 人能写的「关系图」——IR 要序列化的就是这类东西，而不是 task_id=I2V
triples = [
    ("目标视频", "运镜来自", "视频1 / 希区柯克：主体不动镜头推近"),
    ("目标视频", "身份来自", "图2"),
    ("目标视频", "人声来自", "音频3"),
    ("目标视频", "口型对齐", "音频3 的时间轴"),
    ("视频1内部", "镜头类型", "dolly-zoom / 希区柯克"),
]

print("裸 prompt 长度（字）：", len(raw))
print("结构化关系：")
for h, r, t in triples:
    print(f"  {h} --{r}--> {t}")

# 官方口径：素材侧常 100K 理解 token → 平均 4K IR
print("\n压缩比示意：100000 / 4000 =", 100000 / 4000)
print("IR 丢掉的是像素，留下的是「谁和谁什么关系」。像素走 VAE（环节03）。")
```
