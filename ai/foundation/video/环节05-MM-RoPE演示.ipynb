> 验证：相对位移相同 → 注意力分数相同。零依赖。


```python
import math


def rotate(vec2, theta):
    x, y = vec2
    c, s = math.cos(theta), math.sin(theta)
    return (c * x - s * y, s * x + c * y)


def mm_rope(q, k, pos_q, pos_k, base=10000.0):
    tq, hq = pos_q
    tk, hk = pos_k
    th_t = 1.0 / base
    th_h = 1.0 / (base ** 0.5)
    qr = rotate(q[:2], tq * th_t) + rotate(q[2:], hq * th_h)
    kr = rotate(k[:2], tk * th_t) + rotate(k[2:], hk * th_h)
    return sum(a * b for a, b in zip(qr, kr))


q = (1.0, 0.2, 0.5, -0.3)
k = (0.4, 0.8, -0.1, 0.7)
s1 = mm_rope(q, k, (0, 0), (2, 1))
s2 = mm_rope(q, k, (5, 3), (7, 4))
s3 = mm_rope(q, k, (0, 0), (2, 2))
print(f"相对(2,1) @原点  {s1:.6f}")
print(f"相对(2,1) @(5,3) {s2:.6f}")
print(f"相对(2,2)        {s3:.6f}")
assert abs(s1 - s2) < 1e-9
print("相对位移相同 → 分数相同。H3 再多一个 w 轴；音频主要走 t。")
```
