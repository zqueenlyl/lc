> 验证 10 秒 16:9（1344×768）→ 60480 视频 token。零依赖。


```python
def visual_tokens(seconds, height, width, fps=24):
    frames = seconds * fps
    t = frames // 4
    h = (height // 16) // 2
    w = (width // 16) // 2
    return t * h * w, (t, h, w)


n, grid = visual_tokens(10, 768, 1344)
print("默认画幅 1344×768 ≈", 1344 * 768, "像素/帧")
print("patch后网格 (T,H,W) =", grid)
print("视频 token =", n)
assert n == 60480
print("每 token 输入维 24×1×2×2 =", 24 * 2 * 2)
print("音频 token/声道 10×40 =", 10 * 40, "；立体声 packed 可能是 800")
print("视频:每声道音频 ≈", n // 400, ": 1")

n_2k_naive, _ = visual_tokens(10, 768 * 2, 1344 * 2)
print("若空间×2 仍走同一VAE+patch，token ≈", n_2k_naive, "(约×4)")
```
