> Euler：已知直线速度时，从噪声走回 x0。零依赖。


```python
x0 = [1.0, -0.5, 0.25]
eps = [0.2, 0.8, -0.3]
v = [e - a for a, e in zip(x0, eps)]  # dx/dt = v，t:0→1 从数据到噪声


def euler(n_steps):
    x = list(eps)  # t=1
    dt = 1.0 / n_steps
    for _ in range(n_steps):
        x = [xi - dt * vi for xi, vi in zip(x, v)]  # 往 t=0 走
    return x


def mse(a, b):
    return sum((x - y) ** 2 for x, y in zip(a, b)) / len(a)


print("真值", x0)
for n in (1, 2, 8, 32):
    x = euler(n)
    print(f"{n:2d} 步  mse={mse(x, x0):.2e}  { [round(i,4) for i in x] }")
print("直线速度已知时 1 步就精确（矩形积分）。网络 u_θ 有误差时才需要多步。")
print("H3 开源常见约 50 步；序列每步长度不变，不像 LLM 每步 +1 token。")
```
