```python
x0 = [1.0, 0.0, -1.0]
eps = [0.0, 1.0, 2.0]
t = 0.25
xt = [(1 - t) * a + t * e for a, e in zip(x0, eps)]
v = [e - a for a, e in zip(x0, eps)]
print("x0", x0)
print("xt", [round(x, 3) for x in xt])
print("v ", v)
print("检查：xt 是否在线段上", all(abs(xt[i] - ((1-t)*x0[i]+t*eps[i])) < 1e-9 for i in range(3)))


def shift_t(t, s):
    return (s * t) / (1 + (s - 1) * t)


print("\n  t   s=1    音频3   视频12  （工作模型，t=1为噪声；非官方公式）")
for t in [i / 5 for i in range(6)]:
    print(f"{t:3.1f}  {shift_t(t,1):5.3f}  {shift_t(t,3):5.3f}  {shift_t(t,12):5.3f}")

vid_elems = 60480 * 96
aud_elems = 400 * 32
print(f"\n若不加权，视频MSE元素 {vid_elems:,} vs 每声道音频 {aud_elems:,}  ≈ {vid_elems/aud_elems:.0f}×")
```
